```python
LETTERS = ("a", "b", "c", "d")

QUESTIONS = [
    {
        "question": "What is the capital of France?",
        "options": ["London", "Paris", "Berlin", "Madrid"],
        "answer": "b",
    },
    {
        "question": "Which planet is known for its rings?",
        "options": ["Mars", "Venus", "Saturn", "Mercury"],
        "answer": "c",
    },
    {
        "question": "What is the largest ocean on Earth?",
        "options": ["Atlantic", "Indian", "Arctic", "Pacific"],
        "answer": "d",
    },
]

# ---------- INPUT FILTER ----------
def sanitize(text):
    return text.strip().lower()
    
def get_choice():
    while True:
        choice = sanitize(input("Your answer (A/B/C/D):"))     
        if choice in LETTERS:
            return choice
        print("Invalid choice. Please type A, B, C or D.")

# ---------- QUESTION BLOCK (ask, sanitize, evaluate, execute) ----------
def ask_question(number, item):
    print(f"Q{number}. {item['question']}")
    for letter, option in zip(LETTERS, item["options"]):
        print(f"   {letter.upper()}) {option}")

    choice = get_choice()                                      

    if choice == item["answer"]:                              
        print("Correct! +1 point\n")                          
        return True
    else:
        correct_index = LETTERS.index(item["answer"])
        correct_text = item["options"][correct_index]
        print(f"Wrong. The correct answer is {item['answer'].upper()}) {correct_text}\n")
        return False                                         
        
# ---------- OUTPUT ----------
def show_final_score(score, total):
    print("=============================")
    print(f"Final score: {score: > 2} out of {total}")
    print("=============================")

    if score == total:
        print("Perfect score! Excellent work.")
    elif score >= 2:
        print("Good job! Almost there.")
    elif score == 1:
        print("Not bad. Keep practising.")
    else:
        print("Better luck next time. Try again!")

# ---------- MAIN ----------
def main():
    score = 0                     
    total = len(QUESTIONS)

    print("===== GENERAL KNOWLEDGE QUIZ =====")
    print(f"\nAnswer {total} questions. You get +1 point for each correct answer.\n")

    for number, item in enumerate(QUESTIONS, start = 1):
        if ask_question(number, item):
            score += 1             
            
    show_final_score(score, total)
    
if __name__ == "__main__":
    main()
```
